```python
# Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

-  A shallow copy creates a new outer list but shares the inner lists, so modifying an inner list also affects a. 

```python
#Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```

- A deep copy creates independent nested lists, so modifying b does not change a.

```python
#Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

- The copy() method makes a shallow copy, so a and b share their inner lists.

```python
#Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```

-  deepcopy() creates separate inner lists, so appending to a does not affect b.

```python
#Q39
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
```

- The outer lists are different objects, but their first inner list is shared by the shallow copy.

-  *args collects all positional arguments into a tuple.

```python
#Q45
def test(*args):
    print(type(args))

test("a", "b", "c")
```

- The *args parameter stores positional arguments in a tuple, regardless of how many arguments are passed.

```python
#46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
```

- **kwargs collects keyword arguments into a dictionary, using the argument names as keys.

```python
#47
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)
```

- The **kwargs parameter stores keyword arguments in a dictionary.
